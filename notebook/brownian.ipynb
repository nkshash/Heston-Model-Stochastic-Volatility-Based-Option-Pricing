```python
import numpy as np
import plotly.graph_objects as go
import matplotlib.pyplot as plt

```

```python
def brownian_path(T, N):
    """
    Simulates a single path of Brownian motion (Wiener process) over time interval [0, T] with N steps.
    
    Parameters:
    T : float
        Total time duration.
    N : int
        Number of steps.
        
    Returns:
    W : numpy.ndarray
        Simulated Brownian motion path.
    """
    dt = T / N  # Time step
    Z = np.random.normal(0, 1, N)  # Standard normal random variables
    dW = np.sqrt(dt) * Z  # Increments of Brownian motion

    W = np.zeros(N + 1)  # Initialize the Brownian motion array

    for i in range(1, N + 1):
        W[i] = W[i - 1] + dW[i - 1] # Cumulative sum to get the Brownian path

    return W






```

```python


T = 1
N = 1000

paths = []

for i in range(10):
    paths.append(brownian_path(T, N))

W = np.vstack(paths)

x=np.linspace(0, T, N + 1)


fig = go.Figure()

for i in range(W.shape[0]):
    fig.add_trace(go.Scatter(
        x=x,
        y=W[i],
        mode="lines",
        name=f"Path {i + 1}"
    ))

fig.update_layout(
    title="10 Simulated Wiener Paths",
    xaxis_title="Time",
    yaxis_title="Brownian Motion",
    template="plotly_white"
)

fig.show()
```
